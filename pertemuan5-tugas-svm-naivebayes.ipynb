```python
# ============================================================
# 1. CLONE REPOSITORY GITHUB
# ============================================================

from google.colab import drive

drive.mount('/content/drive')

%cd /content/drive/MyDrive

!git clone https://github.com/sofiyah27/pertemuan5-tugas-svm-naivebayes.git

%cd pertemuan5-tugas-svm-naivebayes

!git status


# ============================================================
# 2. IMPORT LIBRARY
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re

from sklearn.model_selection import (
    train_test_split,
    cross_val_score,
    GridSearchCV
)

from sklearn.svm import SVC

from sklearn.naive_bayes import (
    MultinomialNB,
    BernoulliNB
)

from sklearn.feature_extraction.text import (
    CountVectorizer,
    TfidfVectorizer
)

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score
)

from sklearn.pipeline import Pipeline

from wordcloud import WordCloud

np.random.seed(42)

plt.rcParams['figure.figsize'] = (8, 6)

print("Library berhasil diimport!")


# ============================================================
# 3. DATASET
# ============================================================

data = {

    'komentar': [

        # ---------------- POSITIF ----------------

        "Alhamdulillah ceramahnya sangat bermanfaat dan mencerahkan hati",

        "Terima kasih ustadz ilmunya sangat berguna bagi saya",

        "Saya suka penyampaian dakwah yang santun dan jelas",

        "Materi kajian ini mudah dipahami dan sangat menginspirasi",

        "Semoga Allah membalas kebaikan ustadz yang berbagi ilmu",

        "Kajian ini membuat saya semakin semangat beribadah",

        "Penjelasan tentang akhlak sangat bagus dan menenangkan",

        "Kontennya positif serta memberi motivasi untuk berbuat baik",

        "Saya senang mengikuti kajian ini karena banyak manfaat",

        "Dakwahnya disampaikan dengan bahasa sederhana dan sopan",

        "Materi ini menambah wawasan agama saya",

        "Saya merasa termotivasi setelah mendengar ceramah ini",

        "Nasihatnya sangat menyentuh dan membuat hati tenang",

        "Pembahasannya lengkap serta memberikan contoh yang baik",

        "Semoga kajian seperti ini terus rutin dan bermanfaat",


        # ---------------- NEGATIF ----------------

        "Ceramahnya terlalu panjang dan sangat membosankan",

        "Saya tidak paham dengan penjelasan materi tersebut",

        "Suaranya kurang jelas sehingga sulit didengar",

        "Terlalu banyak basa basi dan tidak fokus pada tema",

        "Kontennya tidak relevan dengan kebutuhan saya",

        "Saya kecewa karena penyampaiannya kurang baik",

        "Materinya dangkal dan tidak memberikan penjelasan mendalam",

        "Waktunya terlalu lama sampai membuat saya mengantuk",

        "Tidak ada contoh praktis hanya teori yang membingungkan",

        "Penyampaiannya monoton sehingga saya kehilangan perhatian",

        "Materi terasa berulang dan tidak memberikan hal baru",

        "Suara pembicara kecil dan mengganggu kenyamanan mendengar",

        "Penjelasan terlalu rumit sehingga sulit dipahami",

        "Isi ceramah kurang terstruktur dan membuat bingung",

        "Saya kurang nyaman dengan cara penyampaian dakwah ini",


        # ---------------- NETRAL ----------------

        "Saya baru melihat kajian ini di media sosial",

        "Materinya membahas tentang pentingnya menjaga waktu",

        "Video kajian diunggah pada hari Senin",

        "Ustadz menjelaskan tema sesuai judul yang ditampilkan",

        "Komentar ini hanya menyampaikan informasi tentang kajian",

        "Kajian tersebut dimulai setelah salat Magrib",

        "Materi membahas adab bertetangga dalam kehidupan sehari hari",

        "Saya mencatat judul kajian untuk ditonton nanti",

        "Video ini berdurasi sekitar tiga puluh menit",

        "Pembahasan kali ini berkaitan dengan zakat dan sedekah",

        "Acara kajian dilaksanakan di masjid pesantren",

        "Materi hari ini membahas sejarah perkembangan Islam",

        "Saya melihat jadwal kajian pada unggahan tersebut",

        "Ustadz membahas beberapa poin dalam materi kajian",

        "Video kajian tersedia dan dapat ditonton kembali"
    ],

    'sentimen': (
        ['positif'] * 15 +
        ['negatif'] * 15 +
        ['netral'] * 15
    )
}


df = pd.DataFrame(data)

print("Jumlah data:", len(df))

print("\nDistribusi kelas:")

print(
    df['sentimen'].value_counts()
)


# ============================================================
# 4. MELIHAT DATASET
# ============================================================

display(df)


# ============================================================
# 5. EDA - DISTRIBUSI KELAS
# ============================================================

df['sentimen'].value_counts().plot(
    kind='bar'
)

plt.title(
    'Distribusi Kelas Sentimen'
)

plt.xlabel(
    'Sentimen'
)

plt.ylabel(
    'Jumlah'
)

plt.show()


# ============================================================
# 6. PANJANG TEKS
# ============================================================

df['panjang_teks'] = df['komentar'].apply(
    lambda x: len(x.split())
)

print(
    df[
        [
            'komentar',
            'sentimen',
            'panjang_teks'
        ]
    ].head()
)

df.groupby(
    'sentimen'
)['panjang_teks'].mean().plot(
    kind='bar'
)

plt.title(
    'Rata-rata Panjang Komentar'
)

plt.xlabel(
    'Sentimen'
)

plt.ylabel(
    'Jumlah Kata'
)

plt.show()


# ============================================================
# 7. WORD CLOUD
# ============================================================

semua_teks = ' '.join(
    df['komentar']
)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate(
    semua_teks
)

plt.figure(
    figsize=(12, 6)
)

plt.imshow(
    wordcloud,
    interpolation='bilinear'
)

plt.axis('off')

plt.title(
    'Word Cloud Komentar Dakwah'
)

plt.show()


# ============================================================
# 8. PREPROCESSING
# Case Folding
# Tokenisasi
# Stopword Bahasa Indonesia
# ============================================================

stopwords_id = set("""
yang dan di ke dari ini itu untuk dengan pada saya anda kita mereka adalah akan
dalam tidak sangat sudah sebagai karena atau juga dapat hanya lebih tentang oleh
sebuah ada para agar setelah jika maka bahwa secara menjadi kepada serta seperti
saat terhadap melalui tanpa antara bagi bisa masih telah sedang belum jadi namun
tetapi
""".split())


def preprocessing_teks(teks):

    # Case folding
    teks = teks.lower()

    # Tokenisasi
    tokens = re.findall(
        r'[a-zA-Z]+',
        teks
    )

    # Stopword removal
    tokens = [
        kata
        for kata in tokens
        if kata not in stopwords_id
        and len(kata) > 2
    ]

    return ' '.join(tokens)


df['komentar_bersih'] = df[
    'komentar'
].apply(
    preprocessing_teks
)


display(
    df[
        [
            'komentar',
            'komentar_bersih',
            'sentimen'
        ]
    ].head(10)
)


# ============================================================
# 9. TRAIN TEST SPLIT
# ============================================================

X_train_text, X_test_text, y_train, y_test = train_test_split(

    df['komentar_bersih'],

    df['sentimen'],

    test_size=0.2,

    random_state=42,

    stratify=df['sentimen']
)


print(
    "Jumlah data training:",
    len(X_train_text)
)

print(
    "Jumlah data testing:",
    len(X_test_text)
)


# ============================================================
# 10. COUNTVECTORIZER
# ============================================================

vectorizer = CountVectorizer()

X_counts_train = vectorizer.fit_transform(
    X_train_text
).toarray()

X_counts_test = vectorizer.transform(
    X_test_text
).toarray()

print(
    "Jumlah fitur CountVectorizer:",
    X_counts_train.shape[1]
)


# ============================================================
# 11. TF-IDF VECTORIZER
# ============================================================

tfidf_vectorizer = TfidfVectorizer()

X_tfidf_train = tfidf_vectorizer.fit_transform(
    X_train_text
)

X_tfidf_test = tfidf_vectorizer.transform(
    X_test_text
)

print(
    "Jumlah fitur TF-IDF:",
    X_tfidf_train.shape[1]
)


# ============================================================
# 12. NAIVE BAYES FROM SCRATCH
# MULTINOMIAL + LAPLACE SMOOTHING
# ============================================================

class NaiveBayesScratch:

    def __init__(
        self,
        alpha=1.0
    ):

        self.alpha = alpha

        self.classes = None

        self.prior = {}

        self.likelihood = {}

        self.vocab = None


    def fit(
        self,
        X,
        y
    ):

        self.classes = np.unique(y)

        n_samples, n_features = X.shape

        self.vocab = np.arange(
            n_features
        )


        # Prior probability
        for c in self.classes:

            self.prior[c] = (
                np.sum(y == c)
                /
                n_samples
            )


        # Likelihood
        for c in self.classes:

            X_c = X[
                y == c
            ]

            count_words = np.sum(
                X_c,
                axis=0
            )

            total_words = np.sum(
                count_words
            )

            prob_words = (
                count_words + self.alpha
            ) / (
                total_words
                +
                self.alpha * n_features
            )

            self.likelihood[c] = (
                prob_words
            )


    def predict(
        self,
        X
    ):

        predictions = []


        for x in X:

            posteriors = {}


            for c in self.classes:

                log_posterior = np.log(
                    self.prior[c]
                )

                log_posterior += np.sum(
                    x *
                    np.log(
                        self.likelihood[c]
                    )
                )

                posteriors[c] = (
                    log_posterior
                )


            predictions.append(
                max(
                    posteriors,
                    key=posteriors.get
                )
            )


        return np.array(
            predictions
        )


    def predict_proba(
        self,
        X
    ):

        probas = []


        for x in X:

            log_posteriors = {}


            for c in self.classes:

                log_posterior = np.log(
                    self.prior[c]
                )

                log_posterior += np.sum(
                    x *
                    np.log(
                        self.likelihood[c]
                    )
                )

                log_posteriors[c] = (
                    log_posterior
                )


            max_log = max(
                log_posteriors.values()
            )


            exp_vals = {

                c: np.exp(
                    lp - max_log
                )

                for c, lp
                in log_posteriors.items()
            }


            total = sum(
                exp_vals.values()
            )


            probas.append([

                exp_vals[c] / total

                for c in self.classes

            ])


        return np.array(
            probas
        )


# ============================================================
# 13. PERBANDINGAN ALPHA NAIVE BAYES SCRATCH
# ============================================================

alpha_values = [
    0.01,
    0.1,
    1.0,
    2.0
]

hasil_alpha = []


for alpha in alpha_values:

    nb_scratch = NaiveBayesScratch(
        alpha=alpha
    )

    nb_scratch.fit(
        X_counts_train,
        y_train.values
    )

    y_pred = nb_scratch.predict(
        X_counts_test
    )

    hasil_alpha.append({

        'Alpha': alpha,

        'Accuracy':
            accuracy_score(
                y_test,
                y_pred
            ),

        'F1':
            f1_score(
                y_test,
                y_pred,
                average='weighted'
            )
    })


hasil_alpha_df = pd.DataFrame(
    hasil_alpha
)


print(
    "Hasil perbandingan Alpha:"
)

display(
    hasil_alpha_df
)


# ============================================================
# 14. MODEL LIBRARY
# ============================================================

models = {

    'NB Multinomial':
        MultinomialNB(
            alpha=1.0
        ),

    'NB Bernoulli':
        BernoulliNB(
            alpha=1.0
        ),

    'SVM Linear':
        SVC(
            kernel='linear',
            C=1.0,
            random_state=42
        ),

    'SVM RBF':
        SVC(
            kernel='rbf',
            C=1.0,
            gamma='scale',
            random_state=42
        ),

    'SVM Polynomial':
        SVC(
            kernel='poly',
            C=1.0,
            gamma='scale',
            degree=3,
            random_state=42
        )
}


# ============================================================
# 15. PERBANDINGAN MODEL - COUNTVECTORIZER
# ============================================================

hasil_count = []


for nama, model in models.items():

    model.fit(
        X_counts_train,
        y_train
    )

    y_pred = model.predict(
        X_counts_test
    )


    hasil_count.append({

        'Fitur':
            'CountVectorizer',

        'Model':
            nama,

        'Accuracy':
            accuracy_score(
                y_test,
                y_pred
            ),

        'Precision':
            precision_score(
                y_test,
                y_pred,
                average='weighted',
                zero_division=0
            ),

        'Recall':
            recall_score(
                y_test,
                y_pred,
                average='weighted',
                zero_division=0
            ),

        'F1':
            f1_score(
                y_test,
                y_pred,
                average='weighted'
            )
    })


hasil_count_df = pd.DataFrame(
    hasil_count
)


print(
    "Hasil CountVectorizer:"
)

display(
    hasil_count_df
)


# ============================================================
# 16. PERBANDINGAN MODEL - TF-IDF
# ============================================================

hasil_tfidf = []


for nama, model in models.items():

    model.fit(
        X_tfidf_train,
        y_train
    )

    y_pred = model.predict(
        X_tfidf_test
    )


    hasil_tfidf.append({

        'Fitur':
            'TF-IDF',

        'Model':
            nama,

        'Accuracy':
            accuracy_score(
                y_test,
                y_pred
            ),

        'Precision':
            precision_score(
                y_test,
                y_pred,
                average='weighted',
                zero_division=0
            ),

        'Recall':
            recall_score(
                y_test,
                y_pred,
                average='weighted',
                zero_division=0
            ),

        'F1':
            f1_score(
                y_test,
                y_pred,
                average='weighted'
            )
    })


hasil_tfidf_df = pd.DataFrame(
    hasil_tfidf
)


print(
    "Hasil TF-IDF:"
)

display(
    hasil_tfidf_df
)


# ============================================================
# 17. GABUNGKAN HASIL
# ============================================================

hasil_perbandingan = pd.concat(

    [
        hasil_count_df,
        hasil_tfidf_df
    ],

    ignore_index=True
)


print(
    "Tabel Perbandingan Semua Model:"
)

display(
    hasil_perbandingan
)


# ============================================================
# 18. CLASSIFICATION REPORT SVM LINEAR
# ============================================================

svm_linear = SVC(
    kernel='linear',
    C=1.0,
    random_state=42
)

svm_linear.fit(
    X_counts_train,
    y_train
)

y_pred_svm = svm_linear.predict(
    X_counts_test
)


print(
    classification_report(
        y_test,
        y_pred_svm
    )
)


# ============================================================
# 19. CONFUSION MATRIX MULTI-CLASS
# ============================================================

labels = [
    'negatif',
    'netral',
    'positif'
]


cm = confusion_matrix(

    y_test,

    y_pred_svm,

    labels=labels
)


print(
    "Confusion Matrix:"
)

print(
    cm
)


plt.figure(
    figsize=(7, 6)
)

plt.imshow(
    cm
)

plt.xticks(
    range(len(labels)),
    labels
)

plt.yticks(
    range(len(labels)),
    labels
)

plt.xlabel(
    'Predicted'
)

plt.ylabel(
    'Actual'
)

plt.title(
    'Confusion Matrix SVM Linear'
)


for i in range(
    len(labels)
):

    for j in range(
        len(labels)
    ):

        plt.text(

            j,

            i,

            cm[i, j],

            ha='center',

            va='center'
        )


plt.show()


# ============================================================
# 20. ROC-AUC
# HANYA JIKA DATASET BINARY
# ============================================================

if len(
    np.unique(y_test)
) == 2:

    y_score = svm_linear.decision_function(
        X_counts_test
    )

    print(
        "ROC-AUC:",
        roc_auc_score(
            y_test,
            y_score
        )
    )

else:

    print(
        "ROC-AUC: N/A karena dataset memiliki "
        "3 kelas (multi-class)."
    )


# ============================================================
# 21. 5-FOLD CROSS VALIDATION
# ============================================================

X_counts_all = CountVectorizer().fit_transform(
    df['komentar_bersih']
)


models_cv = {

    'NB Multinomial':
        MultinomialNB(
            alpha=1.0
        ),

    'NB Bernoulli':
        BernoulliNB(
            alpha=1.0
        ),

    'SVM Linear':
        SVC(
            kernel='linear',
            C=1.0,
            random_state=42
        ),

    'SVM RBF':
        SVC(
            kernel='rbf',
            C=1.0,
            gamma='scale',
            random_state=42
        ),

    'SVM Polynomial':
        SVC(
            kernel='poly',
            C=1.0,
            gamma='scale',
            degree=3,
            random_state=42
        )
}


hasil_cv = []


for nama, model in models_cv.items():

    scores = cross_val_score(

        model,

        X_counts_all,

        df['sentimen'],

        cv=5,

        scoring='f1_weighted'
    )


    hasil_cv.append({

        'Model':
            nama,

        'Mean F1':
            scores.mean(),

        'Std F1':
            scores.std()
    })


hasil_cv_df = pd.DataFrame(
    hasil_cv
)


print(
    "Hasil 5-Fold Cross Validation:"
)

display(
    hasil_cv_df
)


# ============================================================
# 22. GRIDSEARCH SVM
# ============================================================

svm_pipeline = Pipeline([

    (
        'tfidf',
        TfidfVectorizer()
    ),

    (
        'svm',
        SVC(
            random_state=42
        )
    )
])


param_grid_svm = {

    'svm__C':
        [0.1, 1, 10, 100],

    'svm__kernel':
        ['linear', 'rbf', 'poly'],

    'svm__gamma':
        [
            'scale',
            'auto',
            0.01,
            0.1
        ]
}


grid_svm = GridSearchCV(

    svm_pipeline,

    param_grid_svm,

    cv=5,

    scoring='f1_weighted',

    n_jobs=-1
)


grid_svm.fit(
    X_train_text,
    y_train
)


print(
    "Parameter SVM terbaik:"
)

print(
    grid_svm.best_params_
)


print(
    "Score CV terbaik:",
    grid_svm.best_score_
)


y_pred_grid_svm = grid_svm.predict(
    X_test_text
)


print(
    "Accuracy SVM GridSearch:",
    accuracy_score(
        y_test,
        y_pred_grid_svm
    )
)


print(
    "F1 SVM GridSearch:",
    f1_score(
        y_test,
        y_pred_grid_svm,
        average='weighted'
    )
)


# ============================================================
# 23. GRIDSEARCH NAIVE BAYES
# ============================================================

nb_pipeline = Pipeline([

    (
        'count',
        CountVectorizer()
    ),

    (
        'nb',
        MultinomialNB()
    )
])


param_grid_nb = {

    'nb__alpha':
        [
            0.01,
            0.1,
            0.5,
            1.0,
            2.0,
            5.0
        ]
}


grid_nb = GridSearchCV(

    nb_pipeline,

    param_grid_nb,

    cv=5,

    scoring='f1_weighted',

    n_jobs=-1
)


grid_nb.fit(
    X_train_text,
    y_train
)


print(
    "Parameter Naive Bayes terbaik:"
)

print(
    grid_nb.best_params_
)


print(
    "Score CV terbaik:",
    grid_nb.best_score_
)


y_pred_grid_nb = grid_nb.predict(
    X_test_text
)


print(
    "Accuracy NB GridSearch:",
    accuracy_score(
        y_test,
        y_pred_grid_nb
    )
)


print(
    "F1 NB GridSearch:",
    f1_score(
        y_test,
        y_pred_grid_nb,
        average='weighted'
    )
)


# ============================================================
# 24. PREDIKSI KOMENTAR BARU
# ============================================================

komentar_baru = [

    "Alhamdulillah kajiannya sangat bermanfaat dan menenangkan hati",

    "Saya kecewa, materinya tidak sesuai dan membosankan",

    "Penyampaian ustadz sangat jelas dan mudah dipahami"
]


prediksi_baru = grid_nb.predict(
    komentar_baru
)


print(
    "Hasil prediksi komentar baru:"
)


for komentar, prediksi in zip(
    komentar_baru,
    prediksi_baru
):

    print(
        "\nKomentar:",
        komentar
    )

    print(
        "Prediksi:",
        prediksi
    )


# ============================================================
# 25. FEATURE IMPORTANCE
# TOP 10 KATA SETIAP KELAS
# ============================================================

best_nb = grid_nb.best_estimator_

nb_model = best_nb.named_steps[
    'nb'
]

count_model = best_nb.named_steps[
    'count'
]

feature_names = count_model.get_feature_names_out()


for i, kelas in enumerate(
    nb_model.classes_
):

    log_prob = nb_model.feature_log_prob_[i]

    top_indices = np.argsort(
        log_prob
    )[-10:][::-1]

    top_words = feature_names[
        top_indices
    ]

    top_values = log_prob[
        top_indices
    ]


    print(
        "\n================================"
    )

    print(
        "Kelas:",
        kelas
    )

    print(
        "10 kata paling indikatif:"
    )

    for word, value in zip(
        top_words,
        top_values
    ):

        print(
            word,
            "->",
            round(value, 4)
        )


    plt.figure(
        figsize=(10, 5)
    )

    plt.barh(
        top_words[::-1],
        top_values[::-1]
    )

    plt.title(
        'Top 10 Kata - ' + kelas
    )

    plt.xlabel(
        'Log Probability'
    )

    plt.ylabel(
        'Kata'
    )

    plt.show()


# ============================================================
# 26. ERROR ANALYSIS
# ============================================================

hasil_error = pd.DataFrame({

    'Komentar':
        X_test_text.values,

    'Aktual':
        y_test.values,

    'Prediksi':
        y_pred_grid_svm

})


error_data = hasil_error[
    hasil_error['Aktual']
    !=
    hasil_error['Prediksi']
]


print(
    "Jumlah data salah:",
    len(error_data)
)


print(
    "Jumlah data testing:",
    len(hasil_error)
)


print(
    "Persentase error:",
    round(
        len(error_data)
        /
        len(hasil_error)
        *
        100,
        2
    ),
    "%"
)


display(
    error_data
)


print(
    "\nKemungkinan perbaikan:"
)

print(
    "1. Menambah jumlah dataset."
)

print(
    "2. Menambah variasi komentar."
)

print(
    "3. Menggunakan preprocessing bahasa Indonesia yang lebih lengkap."
)

print(
    "4. Melakukan tuning parameter model."
)

print(
    "5. Melakukan pemeriksaan ulang terhadap label dataset."
)


# ============================================================
# 27. REFLEKSI ETIKA
# ============================================================

print("""
REFLEKSI ETIKA

Sistem klasifikasi komentar dakwah harus digunakan secara
hati-hati karena komentar masyarakat memiliki konteks yang
beragam.

Dari sisi adl (keadilan), dataset harus memiliki representasi
yang cukup untuk setiap kelas agar model tidak lebih memihak
kepada salah satu kategori.

Dari sisi maslahah, sistem dapat membantu tim media pesantren
menyaring komentar yang perlu diperhatikan. Namun hasil
klasifikasi tidak sebaiknya menjadi keputusan otomatis mutlak.

Komentar yang salah diklasifikasikan dapat menyebabkan
informasi atau pendapat yang sebenarnya baik dianggap negatif.

Karena itu, hasil klasifikasi sebaiknya tetap diperiksa oleh
manusia sebelum digunakan untuk mengambil tindakan terhadap
komentar.
""")


# ============================================================
# 28. HASIL AKHIR
# ============================================================

print("\n==============================================")
print("HASIL AKHIR PRAKTIKUM")
print("==============================================")

print(
    "\nJumlah Dataset:",
    len(df)
)

print(
    "\nKelas:",
    list(
        df['sentimen'].unique()
    )
)

print(
    "\nHasil Perbandingan Model:"
)

display(
    hasil_perbandingan
)

print(
    "\nHasil 5-Fold Cross Validation:"
)

display(
    hasil_cv_df
)

print(
    "\nHasil Alpha Naive Bayes Scratch:"
)

display(
    hasil_alpha_df
)
```
